```python
def generador(a, b, m, x0):
    x1 = (a*x0 + b)%m
    return x1
```

```python
a = 7
b = 5
m = 8
x0 = 1
n = 10

muestra = []
datos = []
for i in range(n):
  nuevo = generador(a, b, m, x0)
  muestra.append([nuevo])
  datos.append(x0)
  x0 = nuevo
print(muestra)
```
