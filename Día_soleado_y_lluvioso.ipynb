#Comportamiento a largo plazo de Cadenas de Markov

Dada la matriz de probabilidades de transición $P$ y el estado inicial %\pi_0$ de una cadena de Markov, la distribución en el tiempo siguiente se calcula como:

$$\pi_1=\pi_0P$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sympy import *
init_printing(use_unicode=True)
```

```python
P=np.array([
    [0.6,0.4],
    [0.2,0.8]
])
```

¿Cuál es la probabilidad de que a largo plazo un día sea soleado? Acompañe su respuesta de una gráfica.

```python
estado = np.array([[1,0]])
```

```python
dias = 30
```

```python
lluvia = []
soleado = []
```

```python
lluvia = []
soleado = []
for n in range(dias + 1):
  lluvia.append(estado[0][0])
  soleado.append(estado[0][1])
  estado = estado @ P
print("probabilidad a largo plazo de que esté soleado", estado[0][1])
print("probabilidad a largo plazo de que llueva", estado[0][0])

plt.figure(figsize=(10,5))
plt.plot(lluvia, label='lluvia')
plt.plot(soleado, label='soleado')
plt.legend()
plt.show()
```
