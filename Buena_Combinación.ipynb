#Cadena de Markov Mediante Caminata Aleatoria Sobre un Grafo de Buenas Secuencias

1.  Primero escogemos una  posición uniformemente al azar entre las $m$ posiciones.
2.   Si en esa posición hay un **1**, la cambiamos a **0**.
3. Si hay **0**, intentamos cambiarla a **1**.
4. Si el cambio produce una buena cadena, nos movemos a esa cadena.
5. Si produce una cadena mala, nos quedamos donde estamos.

De esta forma se cumplen las condiciones de una buena cadena.


```python
import random
import itertools
import networkx as nx
import matplotlib.pyplot as plt
```

```python
def es_buena(cadena):
  for i in range(len(cadena) -1):
    if cadena[i] == 1 and cadena[i + 1] == 1:
            return False

    return True


def caminata_aleatoria(cadena):
    """
    Realiza un paso de la cadena de Markov.
    """

    m = len(cadena)

    # Elegimos una posición uniformemente al azar
    posicion = random.randint(0, m - 1)

    # Hacemos una copia de la cadena
    nueva_cadena = cadena.copy()

    # Si hay un 1, lo cambiamos a 0
    if cadena[posicion] == 1:
        nueva_cadena[posicion] = 0

    # Si hay un 0, intentamos cambiarlo a 1
    else:
        nueva_cadena[posicion] = 1

        # Si deja de ser buena, nos quedamos donde estamos
        if not es_buena(nueva_cadena):
            return cadena, posicion, False

    return nueva_cadena, posicion, True
```

Ahora ejecutamos los pasos de la caminata.

```python
# Longitud de la cadena
m = 10

# Estado inicial: empezamos con una cadena buena
cadena = [0] * m

# Número de pasos de la caminata
pasos = 100
print("Estado inicial:")
print("".join(map(str, cadena)))

for i in range(pasos):

    cadena_anterior = cadena.copy()

    cadena, posicion, movimiento = caminata_aleatoria(cadena)

    print(
        f"Paso {i+1:3d}: "
        f"posición = {posicion}, "
        f"".join(map(str, cadena))
    )
```

```python
# Definimos un valor de m más pequeño para que el grafo sea visualmente legible
m_grafo = 4

# Generamos todas las cadenas binarias posibles de longitud m_grafo
todas_las_cadenas = [list(x) for x in itertools.product([0, 1], repeat=m_grafo)]

# Filtramos para quedarnos solo con las buenas secuencias (nodos del grafo)
nodos_buenos = [tuple(cadena) for cadena in todas_las_cadenas if es_buena(cadena)]

# Creamos el grafo
G = nx.Graph()
G.add_nodes_from(nodos_buenos)

# Añadimos aristas entre secuencias que difieren en exactamente 1 posición
for i in range(len(nodos_buenos)):
    for j in range(i + 1, len(nodos_buenos)):
        nodo1 = nodos_buenos[i]
        nodo2 = nodos_buenos[j]
        # Contamos en cuántas posiciones difieren
        diferencias = sum(1 for a, b in zip(nodo1, nodo2) if a != b)
        if diferencias == 1:
            G.add_edge(nodo1, nodo2)

# Calculamos las posiciones para dibujar el grafo
pos = nx.spring_layout(G, seed=42)

# Ajustamos las variables de la simulación para que coincidan con este nuevo grafo
m = m_grafo
cadena = [0] * m
pasos = 10  # Reducimos los pasos de visualización para no generar 100 imágenes

print(f"Grafo creado con {G.number_of_nodes()} nodos (buenas secuencias) y {G.number_of_edges()} aristas.")
```
