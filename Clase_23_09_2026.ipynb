```python
import matplotlib.pyplot as plt
import numpy as np
import random as r
```

```python
def f(x, lam):
  return lam*np.exp(-lam*x)
```

```python
x = np.linspace(-10,10,100)
plt.plot(f(x,1))
```

```python
def F(x,lam):
  return 1-np.exp(-lam*x)
plt.plot(F(x,1))
```
